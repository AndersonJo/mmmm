```python
d = 256 # Dimension size
n_samples = 10_000
a = np.random.randn(n_samples, d)   # shape: (n_samples, d)
b = np.random.randn(n_samples, d)
print('a:', a.shape)
print('b:', b.shape)

e = (a * b).sum(axis=1)

print('mean:', e.mean())
print('var:', e.var())
print('std:', e.std())
```
