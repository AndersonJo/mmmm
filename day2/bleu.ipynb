```python
hyp = "the cat sat on the mat".split()
ref = "the cat is on the mat".split()

for n in range(1, 5):
    p = modified_precision(hyp, [ref], n)
    print(f'{n}-gram precision: {p:.3f}')

plot_ngram_bars(hyp, ref)
```

```python
# "the" 를 6번 반복해도 클리핑 덕분에 precision 이 낮게 유지됨
cheat = "the the the the the the".split()
print('클리핑 전 (단순 카운트):', 6/6)     # 6/6 = 1.0 (부당)
print('클리핑 후 (modified): ', modified_precision(cheat, [ref], 1))  # 2/6
print('정상 가설:             ', modified_precision(hyp, [ref], 1))
```
