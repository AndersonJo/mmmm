# Luong

```python
rng = np.random.default_rng(0)
s = rng.standard_normal(8)  # 상태 s
H = rng.standard_normal((5, 8))  # hidden states H
score_dot = H @ s

print(s.shape, H.shape)
print(score_dot)  # softmax numpy  찾아서 score_dot 넣어보자
```
