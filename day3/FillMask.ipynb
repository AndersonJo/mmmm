```python
name = 'klue/bert-base'
tok = load_tokenizer(name)
model = load_model(name, task='mlm')
model.eval()
print(tok.mask_token, tok.mask_token_id)
```

```python
mask_token = tok.mask_token
sent = f'한국의 수도는 {mask_token} 입니다.'
topk = predict_mask_topk(model, tok, sent, k=5)
for t, s in topk:
    print(f'{t}: {round(s, 3)}')

# 의사는 병원에서 ___ 을(를) 만났다
# 간호사는 병원에서 ___ 을(를) 만났다
# ___  는 훌륭한 과학자이다
# 그녀는 화가 나서 ___ 을(를) 했다
# ___ 는 집에서 요리를 하고 아이를 돌볼다
# 그녀는 화가 나서 ___ 했다
# 그는 화가 나서 ___ 했다
# 외국인 노동자는 보통 ___에서 일한다
# 그 사람은 ___ 출신이라 성격이 거칠다
```
