# Token

"음식" => 53424
"바나나" => 22554

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained('klue/bert-base')

print('Tokenizer Type:', type(tokenizer).__name__)
print('Vocab:', tokenizer.vocab_size)
```

```python
sentence = '하이닉스 주가가 오늘 박살.. 하지만 우리는 공부를 합니다'

tokens = tokenizer.tokenize(sentence)
print('sentence:', sentence)
print(tokens)
```

```python
encoded = tokenizer(sentence, return_tensors='pt')

print(encoded['input_ids'])
print(encoded['attention_mask'])
print(tokenizer.cls_token_id)  # "[CLS]"
print(tokenizer.sep_token_id)  # "[SEP]"
# "[CLS]하이닉스 반도체는 좋습니다. [SEP]던킨도너츠 어제 먹었는데 맛있다"
```

```python
# Decode
ids = encoded['input_ids'][0].tolist()
print(ids)
decoded = tokenizer.decode(ids, skip_special_tokens=False)
print(decoded)
```

# 신조어를 찾아서 넣어보기

```python
rare_words = [
    '트랜스포머'
]
for word in rare_words:
    toks = tokenizer.tokenize(word)
    print(toks)
```

```python
sentences = [
    '하이닉스 주가가 오늘 많이 빠졌지만 딥러닝은 재미지다',
    '안녕 대한민국',
    # 넣어보세요
]
batch = tokenizer(sentences, padding=True, truancation=True, max_length=32,
         return_tensors='pt')
#  for loop 돌려서 값 찍어보세요
for i, s in enumerate(sentences):
    ids = batch['input_ids'][i].tolist()
    print(ids)
```
