```python
from tool1  import *
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
set_seed(42)
```

```python
class ToyDataset(Dataset):
    def __init__(self, size=1000):
        self.x = torch.randn(size, 4)
        self.y = (self.x.sum(dim=1) > 0).long()
    
    def __len__(self):
        return len(self.x)
    
    def __getitem__(self, idx):
        return self.x[idx], self.y[idx]
dataset = ToyDataset(1000)
print('dataset size:', len(dataset))
print('end=dataset[0]:', dataset[0])
```

```python
loader = DataLoader(dataset, batch_size=64, shuffle=True)
for xb, yb in loader:
    print('xb:', xb.shape, yb.shape)
    break
print('total_batch:', len(loader))
```

```python
# Shuffle
small = ToyDataset(20)
loader_off = DataLoader(small, batch_size=4, shuffle=False)
loader_on = DataLoader(small, batch_size=4, shuffle=True)

print('First batch with shuffle=False:')
for xb, yb in loader_off:
    print(yb.tolist())

print('First batch with shuffle=True:')
for xb, yb in loader_on:
    print(yb.tolist())
```
