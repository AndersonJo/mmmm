```python
w = torch.tensor(2.0, requires_grad=True)
x = torch.tensor(3.0)

y = w * x + 1.0
loss = (y - 10.0) ** 2
loss.backward()
print('w.grad:', w.grad.item(), 'excepted:', 2 * (w.item() * x.item() + 1.0 - 10.0) * x.item())
```
