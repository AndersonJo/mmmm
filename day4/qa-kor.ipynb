```python
from tool3_2 import *
from transformers import AutoTokenizer, AutoModelForQuestionAnswering
```

```python
MODEL_NAME = 'monologg/koelectra-base-v3-finetuned-korquad'

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
qa_model = AutoModelForQuestionAnswering.from_pretrained(MODEL_NAME)
qa_model.eval()
```

```python
context_history = """
조선왕조는 1392년 이성계가 고려를 무너뜨리고 건국한 나라입니다.
수도를 개성에서 한양(현재의 서울)으로 옮겼으며, 약 519년간 지속되었습니다.
세종대왕은 조선의 4대 왕으로, 1443년 훈민정음(한글)을 창제하였습니다.
임진왜란은 1592년 일본이 조선을 침략한 전쟁으로, 이순신 장군이 큰 활약을 했습니다.
조선은 1897년 대한제국으로 국호를 바꾸었고, 1910년 일제에 병합되었습니다.
""".strip()
question1 = '조선왕조를 건국한 사람은 누구입니까?'
answer = qa_pipeline_09_qa_extractive(question=question1, 
                                      context=context_history, 
                                      tokenizer=tokenizer, 
                                      qa_model=qa_model)
print(answer)
```

```python
questions = [
    "한글을 창제한 왕은 누구입니까?",
    "훈민정음은 몇 년에 만들어졌습니까?",
    "Nvidia 가격은?"]
for q in questions:
    ans = qa_pipeline_09_qa_extractive(question=q, context=context_history, tokenizer=tokenizer, qa_model=qa_model)
    print(f'Q: {q}')
    print(f'A: {ans["answer"]}  (신뢰도: {ans["score"]:.1%})')
    print()
```
